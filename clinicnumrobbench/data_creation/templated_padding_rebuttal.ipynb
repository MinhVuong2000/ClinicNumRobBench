# Create padding_templated data

For demonstrating that the big "robustness gap” does not primarily reflect token budget limitations.

```python
import os
import pandas as pd


templated_data_dir = "templated"
new_templated_padding_data_dir = "padding_templated"

for task in os.listdir(templated_data_dir):
    new_dir = os.path.join(new_templated_padding_data_dir,task)
    os.makedirs(new_dir, exist_ok=True)
    for sub_task in os.listdir(os.path.join(templated_data_dir,task)):
        df = pd.read_csv(os.path.join(templated_data_dir,task,sub_task))
        df['context'] = df['question'].map(lambda x: "\n".join(x.split('\n\n')[:-1]).strip())
        df['question'] = df['question'].map(lambda x: x.split('\n\n')[-1].strip())#.split('?')[0].strip())
        df['context'] = df['context'].map(lambda q: q+(q.split('.')[0]+". ")*38)
        df['question'] = df.apply(lambda r: "\n\n".join([r["context"],r['question']]), axis=1)
        df.drop(columns=["context"], inplace=True)
        df.to_csv(os.path.join(new_dir,sub_task), index=False)
```
